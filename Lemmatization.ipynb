# Lemmatization
-  Just like stemming but the output we get after lemmatization is a root word e.g: eaten --> eat

# **WordNet Lemmatizatizer**

```python
from nltk.stem import WordNetLemmatizer
```

```python
lemmatizer=WordNetLemmatizer()
```

```python
lemmatizer.lemmatize("going",pos="n")


# pos n -> noun
# v -->verb
# a -> adjective
# r --> adverb
```

```python
words=["eat","eaten","eating","program","programming","Write","writing"]
```

```python
for word in words:
    print(word+"-->"+lemmatizer.lemmatize(word,pos="n"))
```
