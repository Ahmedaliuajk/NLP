# Stemming
- Reducing a word to its word stem

```python
words=["eat","eaten","eating","program","programming","Write","writing"]
```

## PorterStemmer

```python
from ntlk.stem import PorterStemmer
```

```python
Stemming=PorterStemmer()
```

```python
for word in words:
    print(word+"-->"+Stemming.stem(word))
```

```python
Stemming.stem("congratulations")
Stemming.stem("sitting")
```

## RegexpStemmer Class

- It basically takes a single regular expression and remove prefix or suffix that matches the expression

```python
from nltk.stem  import RegexpStemmer
```

```python
reg_stemmer=RegexpStemmer("ing$ | s$| es$| able$", min=4)

# Dollar sign ($) will remove the suffix if there is no dollar then prefix is removed
```

```python
reg_stemmer.stem("eating")
```

# Snowball Stemmer

```python
from nltk.stem import SnowballStemmer
```

```python
snowballsstemer=SnowballStemmer("english")
```

```python
for word in words:
    print(word+"-->"+snowballsstemer.stem(word))
```
