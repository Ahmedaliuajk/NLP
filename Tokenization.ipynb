```python
corpus="""
Hi, My name is Ahmed.
I am from Muzaffarabad Azad Kashmir.
I am currently doing Bachelors of Software Engineering
"""
```

```python
print(corpus)
```

# Tokenization
## paragraph --> sentences

```python
from nltk.tokenize import sent_tokenize

```

```python
documents=sent_tokenize(corpus)
```

```python
documents
```

```python
type(documents)
```

```python
for sentence in documents:
    print(sentence)
```

# Tokenization
## paragraph --> words
## sentence --> words

```python
from nltk.tokenize import word_tokenize
```

```python
word_tokenize(corpus)
```

```python
from nltk.tokenize import wordpunct_tokenize 

# It also separates the punctuation
```

```python
from nltk.tokenize import TreeBankWordTokenizer

tokenizer=TreeBankWordTokenizer()
tokenizer.tokenize(corpus)  

# in above full stop is included in word
```
